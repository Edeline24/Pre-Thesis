```python
!pip install groq datasets pandas matplotlib -q

from google.colab import userdata
from groq import Groq
from collections import Counter
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt
import time
import math
import re
```

```python
API_KEY = userdata.get('GROQ_API_KEY')

client = Groq(api_key=API_KEY)

MODEL_NAME = "llama-3.3-70b-versatile"

ds = load_dataset("gsm8k", "main", split="test")

print(f"Dataset loaded: {len(ds)} questions")

pilot_ds = ds.select(range(10))
print(f"Pilot dataset loaded: {len(pilot_ds)} questions")
```

```python
def build_prompt(question, options=None):
  # For gsm8k, options are not used, so we create a direct prompt for problem solving.
  prompt = f"""
Question: {question}

Let's think step by step carefully and write out the solution.
Finally, give your answer in the format: The answer is X.
"""

  return prompt
```

```python
def generate_response(prompt, temp=0.7):
  response = client.chat.completions.create(
      model=MODEL_NAME,
      messages=[
          {"role": "user", "content": prompt},
      ],
      temperature=temp,
      max_tokens=400
  )
  return response.choices[0].message.content
```

```python
def extract_answer(response):
  # For gsm8k, we need to extract a numerical answer.
  # Look for 'The answer is X' or '#### X'
  match = re.search(r"The answer is\s*(\-?\d+(?:\.\d+)?)", response, re.IGNORECASE)
  if match:
    return match.group(1)
  match = re.search(r"####\s*(\-?\d+(?:\.\d+)?)", response)
  if match:
    return match.group(1)
  # Fallback: try to find the last number in the response
  numbers = re.findall(r"\-?\d+(?:\.\d+)?", response)
  if numbers:
    return numbers[-1]
  return None
```

```python
def calc_mar(answers):
  if len(answers) == 0:
    return 0

  counter = Counter(answers)
  majority_count = counter.most_common(1)[0][1]
  mar = majority_count / len(answers)
  return mar
```

```python
def calc_entropy(answers):
  if len(answers) == 0:
    return 0

  counter = Counter(answers)
  total = len(answers)
  entropy = 0

  for count in counter.values():
    prob = count / total
    entropy += -prob * math.log2(prob)
  return entropy
```

```python
def extract_numerical_answer_from_gsm8k(answer_string):
  """Extracts the final numerical answer from gsm8k's 'answer' field."""
  # Pattern for '#### X'
  match = re.search(r"####\s*(\-?\d+(?:\.\d+)?)", answer_string)
  if match:
    try:
      return float(match.group(1))
    except ValueError:
      return None
  # Fallback to finding the last number if '####' isn't present
  numbers = re.findall(r"\-?\d+(?:\.\d+)?", answer_string)
  if numbers:
    try:
      return float(numbers[-1])
    except ValueError:
      return None
  return None


results = []
NUM_SAMPLES = 5
print("\nStarting Pilot Experiment...\n")

for index, item in enumerate(pilot_ds):
  print("=" * 60)
  print(f"Question {index + 1}/{len(pilot_ds)}")
  print("=" * 60)

  question = item["question"]
  options = [] # GSM8K does not have options for multiple choice

  ground_truth_full = item["answer"]
  ground_truth_numeric = extract_numerical_answer_from_gsm8k(ground_truth_full)

  print(f"Ground Truth (full): {ground_truth_full}")
  print(f"Ground Truth (numeric): {ground_truth_numeric}")

  responses = []
  extracted_answers_numeric = [] # Store numeric extracted answers

  for i in range(NUM_SAMPLES):
    print(f"\nSample {i + 1}/{NUM_SAMPLES}")

    prompt = build_prompt(question, options)
    response = generate_response(
        prompt,
        temp=0.7
      )

    extracted_answer_str = extract_answer(response)
    responses.append(response)

    if extracted_answer_str is not None:
      try:
        extracted_answers_numeric.append(float(extracted_answer_str))
      except ValueError:
        pass # Ignore non-numeric extractions

    print(f"Extracted Answer: {extracted_answer_str}")

    time.sleep(1)

  most_common_numeric = None
  if extracted_answers_numeric:
    counter = Counter(extracted_answers_numeric)
    most_common_numeric = counter.most_common(1)[0][0]

  # MAR and Entropy are not directly applicable for GSM8K numerical answers, set to 0
  mar = 0.0
  entropy = 0.0

  is_correct = False
  if most_common_numeric is not None and ground_truth_numeric is not None:
    # Compare floating point numbers with a small tolerance
    is_correct = (abs(most_common_numeric - ground_truth_numeric) < 1e-6)

  results_item = {
      "question": question,
      "ground_truth_full": ground_truth_full,
      "ground_truth_numeric": ground_truth_numeric,
      "responses": responses,
      "extracted_answers_numeric": extracted_answers_numeric,
      "most_common_numeric": most_common_numeric,
      "mar": mar,
      "entropy": entropy,
      "correct": is_correct
  }
  results.append(results_item)

  print("\nRESULT")
  print("-" * 40)
  print(f"Extracted Numeric Answers: {extracted_answers_numeric}")
  print(f"Most Common Numeric: {most_common_numeric}")
  print(f"Correct: {is_correct}")
```

```python
print("\n")
print("=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

acc = sum([r["correct"] for r in results]) / len(results)
# MAR and Entropy are not applicable for GSM8K, so these will be 0 as set in the loop.
avg_mar = sum([r["mar"] for r in results]) / len(results)
avg_entropy = sum([r["entropy"] for r in results]) / len(results)

print(f"Accuracy: {acc:.2%}")
print(f"Average MAR (not applicable): {avg_mar:.3f}")
print(f"Average Entropy (not applicable): {avg_entropy:.3f}")

df = pd.DataFrame([
    {
        "Question": r["question"],
        "Ground Truth Numeric": r["ground_truth_numeric"],
        "Most Common Numeric": r["most_common_numeric"],
        "Correct": r["correct"]
    } for r in results
])

print("\nDataFrame Preview:")
print(df.head())

# The original plots for MAR and Agreement vs Correctness are less relevant now.
# We can plot Accuracy or distribution of extracted answers if needed.

# Plotting example for numerical correctness
plt.figure(figsize=(8, 5))
correct_counts = df["Correct"].value_counts().sort_index()
correct_counts.plot(kind='bar', color=['red', 'green'])
plt.title('Correct vs Incorrect Predictions')
plt.xlabel('Prediction Correctness')
plt.ylabel('Number of Questions')
plt.xticks(ticks=[0, 1], labels=['Incorrect', 'Correct'], rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

df.to_csv("pilots_result.csv", index=False)

print("\nResults saved")
```
